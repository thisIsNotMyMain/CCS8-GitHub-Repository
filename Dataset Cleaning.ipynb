```python
import pandas as pd
from google.colab import drive

drive.mount('/content/drive')
df = pd.read_csv(
    "/content/drive/MyDrive/Philippine_Earthquakes_USGS.csv"
)
```

```python
# Step 1
# remove: rows with blank magnitude
# remove: rows with 0.0 magnitude
# remove: rows whose magType is not 'mb'
# fill: rms (median)
# fill: depth (median)
# 1. remove rows with blank (NaN) magnitude
df = df.dropna(subset=["mag"])

# 2. remove rows with 0.0 magnitude
df = df[df["mag"] != 0.0]

# 3. keep only rows where magType == 'mb'
df = df[df["magType"] == "mb"]

# 4. fill rms with median
df["rms"] = df["rms"].fillna(df["rms"].median())

# 5. fill depth with median
df["depth"] = df["depth"].fillna(df["depth"].median())

# optional: reset index after filtering
df = df.reset_index(drop=True)
```

```python
# Step 2:
# Normalize: depth, sig, rms; use Z-score standardization
# Normalize: longitude, latitude; use min-max scaling
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# ---- Z-score standardization ----
z_cols = ["depth", "sig", "rms"]

z_scaler = StandardScaler()
df[z_cols] = z_scaler.fit_transform(df[z_cols])

# ---- Min-max scaling (0 to 1) ----
mm_cols = ["longitude", "latitude"]

mm_scaler = MinMaxScaler()
df[mm_cols] = mm_scaler.fit_transform(df[mm_cols])
```

```python
# Step 3
# drop
cols_to_drop = [
    "title", "nst", "updated", "tz", "url", "detail",
    "felt", "cdi", "mmi", "alert", "magType",
    "status", "tsunami", "net", "code",
    "ids", "sources", "types", "type_property", "gap", "dmin"
]
df = df.rename(columns={"mag": "mag(mB)"})
df = df.drop(columns=cols_to_drop, errors="ignore")
```
